# ავატარის ლურჯი წრის ტესტი

ამ notebook-ში შევამოწმებთ, როგორ გამოიყურება ავატარის ძირითადი ლურჯი წრე matplotlib-ის გამოყენებით.

## საჭირო ბიბლიოთეკების იმპორტი

შემოვიტანოთ matplotlib და numpy, რათა შევძლოთ წრის დახატვა.

```python
import matplotlib.pyplot as plt
import numpy as np
```

## ლურჯი წრის დახატვა matplotlib-ით

ამ სექციაში დავხატავთ ლურჯ წრეს, რომელიც ავატარის ძირითადი ფორმაა.

```python
def draw_blue_avatar_circle(radius=1.0, color='#2196F3'):
    fig, ax = plt.subplots(figsize=(4, 4))
    circle = plt.Circle((0, 0), radius, color=color, ec='black', lw=2)
    ax.add_patch(circle)
    ax.set_xlim(-1.2, 1.2)
    ax.set_ylim(-1.2, 1.2)
    ax.set_aspect('equal')
    ax.axis('off')
    plt.show()
```

## ავატარის ტესტის გაშვება და შედეგის ჩვენება

გავუშვათ ფუნქცია და ვნახოთ, როგორ გამოიყურება ავატარის ლურჯი წრე.

```python
draw_blue_avatar_circle()
```
